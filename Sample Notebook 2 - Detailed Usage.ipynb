```python
from mrm_template_audit.audit_mrm_doc import audit_model_document

import pandas as pd
pd.set_option('display.max_colwidth', None)

# 1. Define your file paths
MODEL_DOC = './test/MDD_Submission_v2 - TOC.docx'
TEMPLATE_DOC = './test/2_MDD_Template_TOC_NoManual_Section_Number.docx'

# 2. Call the audit function
# This assumes get_numbering_definitions, get_numbering_info, 
# extract_toc_from_document, and extract_heading_structure_with_numbering 
# are all defined in your script.
results = audit_model_document(MODEL_DOC, TEMPLATE_DOC)

# 3. Process and Display the Internal Integrity Report
print("=== INTERNAL INTEGRITY AUDIT (TOC vs. Body) ===")
if not results["internal_discrepancies"]:
    print("No discrepancies found. TOC matches the Document Body.")
else:
    df_internal = pd.DataFrame(results["internal_discrepancies"])
    display(df_internal)

print("\n" + "="*50 + "\n")

# 4. Process and Display the Template Compliance Report
print("=== TEMPLATE COMPLIANCE AUDIT (Model vs. Standard) ===")
if not results["template_discrepancies"]:
    print("Document is 100% compliant with the Enterprise Template.")
else:
    df_template = pd.DataFrame(results["template_discrepancies"])
    # Highlight missing sections vs numbering errors
    display(df_template)
```
